```python
import moomoo as ft

ctx = ft.OpenQuoteContext(host="127.0.0.1", port=11111)

for p in [
    ft.PeriodType.INTRADAY,
    ft.PeriodType.DAY,
    ft.PeriodType.WEEK,
    ft.PeriodType.MONTH,
]:
    print("=" * 60)
    print("period =", p)

    ret, data = ctx.get_capital_flow(
        "HK.00700",
        period_type=p,
        start="20260101",
        end="20260714",
    )

    print("ret =", ret)
    print("type =", type(data))
    print(data)

ctx.close()
```
